## **06.1. JOINs and UNIONs**

### **06.1.1. JOINs**


- `(INNER) JOIN`: Devuelve únicamente las filas que tienen valores coincidentes en ambas tablas.
- `LEFT (OUTER) JOIN`: Devuelve todas las filas de la tabla de la izquierda y solo las filas coincidentes de la tabla de la derecha.
- `RIGHT (OUTER) JOIN`: Devuelve todas las filas de la tabla de la derecha y solo las filas coincidentes de la tabla de la izquierda.
- `FULL (OUTER) JOIN`: Devuelve todas las filas cuando hay una coincidencia en la tabla de la izquierda o en la de la derecha.

![Imagen de los tipos de JOINs](./images/JOINsTypes.png)

```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("Chinook_Sqlite.sqlite")

# Realizar tu consulta SQL
query = """
SELECT ar.Name, al.Title
FROM Artist AS ar
INNER JOIN Album AS al
ON ar.ArtistId = al.ArtistID;
"""

db = pd.read_sql_query(query, conexion)

display(db.loc[:10,:])

# Cerrar la conexión al terminar
conexion.close()
```

### **06.1.2. UNIONs**
El operador *UNION* se utiliza para combinar el conjunto de resultados de dos o más sentencias *SELECT*.

El operador *UNION* elimina automáticamente las filas duplicadas del conjunto de resultados.

Requisitos para el uso de *UNION*:
- Todas las sentencias *SELECT* incluidas en *UNION* deben tener el mismo número de columnas.
- Las columnas también deben tener tipos de datos similares.
- Las columnas de todas las sentencias *SELECT* deben aparecer en el mismo orden.


```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("Chinook_Sqlite.sqlite")

# Realizar tu consulta SQL
query = """
SELECT Composer FROM Track
UNION
SELECT Title FROM Album;
"""

df = pd.read_sql_query(query, conexion)

# Mostrar las primeras filas
display(df.iloc[:10,:])

# Cerrar la conexión al terminar
conexion.close()
```

##### **06.1.2.1. UNION ALL**
El operador *UNION* ALL se utiliza para combinar el conjunto de resultados de dos o más sentencias *SELECT*.

El operador *UNION* ALL incluye todas las filas de cada sentencia, incluidos los posibles duplicados.

Requisitos para *UNION* ALL:
- Todas las sentencias *SELECT* incluidas en *UNION* ALL deben tener el mismo número de columnas.
- Las columnas también deben tener tipos de datos similares.
- Las columnas de todas las sentencias *SELECT* deben estar, además, en el mismo orden.


```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("Chinook_Sqlite.sqlite")

# Realizar tu consulta SQL
query = """
SELECT Composer FROM Track
UNION ALL
SELECT Title FROM Album;
"""

df = pd.read_sql_query(query, conexion)

# Mostrar las primeras filas
display(df.iloc[:10,:])

# Cerrar la conexión al terminar
conexion.close()
```

## **06.2. Funciones analiticas**
A continuación se explciara como definir funciones analíticas, que también operan sobre un conjunto de filas. Sin embargo, a diferencia de las funciones de agregación, las funciones analíticas devuelven un valor (que puede ser diferente) para cada fila de la tabla original.

Las funciones analíticas nos permiten realizar cálculos complejos con una sintaxis relativamente sencilla. Por ejemplo, podemos calcular rápidamente medias móviles y totales acumulados, entre otras magnitudes.

### **06.2.1. Sintaxis**
Creación de una base de datos para usarla en este curso.

```python
import pandas as pd
import sqlite3

# 1. Conectar a la base de datos de SQLite
conn = sqlite3.connect("baseDeDatosPrueba.db")
cursor = conn.cursor()

# 2. Ejecutar SQL para crear la tabla
cursor.execute("""
CREATE TABLE IF NOT EXISTS usuarios (
    id INTEGER,
    date DATE,
    time INTEGER
);
""")

# 3. Ejecutar SQL para insertar los registros
cursor.execute("""
INSERT INTO usuarios (id, date, time) VALUES
    (1, '2019-07-05', 22),
    (1, '2019-04-15', 26),
    (2, '2019-02-06', 28),
    (1, '2019-01-02', 30),
    (2, '2019-08-30', 20),
    (2, '2019-03-09', 22);
""")

# Confirmar y guardar los cambios realizados
conn.commit()

# 4. Exportar toda la base de datos a un archivo ejecutable .sql
with open("mi_base_de_datos_dump.sql", "w", encoding="utf-8") as f:
    for linea in conn.iterdump():
        f.write(f"{linea}\n")

# 5. Consultar la base de datos mediante SQL y cargar el resultado en Pandas
query = "SELECT * FROM usuarios"
df = pd.read_sql_query(query, conn)

display(df)

# Cerrar la conexión
conn.close()
```

# **09. Ingenieria de características (Feature Engineering)**


## **09.1. Introducción**
El objetivo de la ingenieria de características es simplemente hacer que los datos esten mejor adaptados al problema que se esta atacando.

Se usa esto para mejorar el rendimiento de la predictividad del modelo, reducir la necesidad computacional y de datos, y mejorar el interpretabilidad de los resultados.

### **09.1.1. El principio de guia de la ingenieria de características**
Para que una característica sea útil, debe tener una relación con la variable de respuesta que tu modelo sea capaz de aprender. Los modelos lineales, por ejemplo, solo son capaces de aprender relaciones lineales. Por lo tanto, cuando utilizas un modelo lineal, tu objetivo es transformar las características para que su relación con la variable de respuesta sea lineal.

La idea clave aquí es que una transformación que se aplica a una característica se convierte, en esencia, en parte del propio modelo. Supongamos que intentas predecir el precio de unas parcelas cuadradas a partir de la longitud de uno de sus lados. Ajustar un modelo lineal directamente a la longitud da malos resultados: la relación no es lineal.

![imagen](./images/principioDeGuiaDeLaIngenieriaDeCaracteristicas1Tabla.png)

Teniendo en cuenta esta tabla se pude observar que el modelo lineal no coincide de forma correcta con los datos por lo tanto su rendimiente se vera afectado, asi que si cuadratrizamos la longitud que esta marcada en el diagrama como ''Lenght'' para conserguir que el se adecue al area ya que se crea una parabo.la, en otras palabras se puede dar a un modelo lineal la habilidad de adaptarse a rasgos cuadraticos.

## **09.2. Información mutua**

### **09.2.1. ¿Qué es la información mutua y que mide?**
Es la medición entre dos cantidades de las cuales con el conocimiento de una de las cantidades reduce el desconocimiento de la otra cantidad.

### **09.2.2. Interpretración de los resultados de la información mutua**
La información mutua mínima posible entre dos magnitudes es 0,0. Cuando la información mutua (MI->Mutual Information) es cero, las magnitudes son independientes: ninguna de ellas puede aportar información sobre la otra. Por el contrario, en teoría no existe un límite superior para el valor que puede alcanzar la MI. Sin embargo, en la práctica, los valores superiores a 2,0 aproximadamente son poco habituales. (La información mutua es una magnitud logarítmica, por lo que aumenta muy lentamente).

La siguiente figura te dará una idea de cómo se corresponden los valores de la información mutua con el tipo y el grado de asociación que tiene una característica con la variable objetivo.

![imagen](./images/mutualInformationTable.png)

A continuación se indican algunos aspectos que hay que tener en cuenta al aplicar la información mutua:

- La información mutua (MI) puede ayudarte a comprender el potencial relativo de una característica como predictor de la variable objetivo, considerada por sí sola.
- Es posible que una característica sea muy informativa cuando interactúa con otras características, pero no lo sea tanto por sí sola. La MI no puede detectar interacciones entre características. Se trata de una métrica univariante.
- La utilidad real de una característica depende del modelo con el que se utilice. Una característica solo es útil en la medida en que su relación con la variable objetivo sea algo que tu modelo pueda aprender. El hecho de que una característica tenga una puntuación alta en la información mutua no significa que tu modelo vaya a poder hacer algo con esa información. Es posible que tengas que transformar primero la característica para poner de manifiesto la asociación.


## **09.3. Agrupación (Clustering) por K-Means**
Mide la similitud utilizando la distancia en línea recta habitual (es decir, la distancia euclidiana). Crea agrupaciones situando una serie de puntos, denominados <ins>centroides</ins>, dentro del espacio de características. A cada punto del conjunto de datos se le asigna la agrupación del centroide al que se encuentre más cerca. La ''k'' de ''k-means'' indica el número de centroides (es decir, de agrupaciones) que se crean. El valor de k lo defines tú mismo.

Podrías imaginarte que cada centroide capta puntos a través de una secuencia de círculos radiales. Cuando los conjuntos de círculos de los centroides competidores se solapan, forman una línea. El resultado es lo que se denomina un mosaico de Voronoi. El mosaico muestra a qué agrupaciones se asignarán los datos futuros; el mosaico es, en esencia, lo que el método k-means aprende de sus datos de entrenamiento.

La agrupación en el conjunto de datos de Ames anterior es una agrupación k-means. Aquí se muestra la misma figura con la teselación y los centroides.

![imagenes](./images/k-meansAgrupacion.jpg)

Repasemos cómo el algoritmo k-means aprende los clústeres y qué significa eso para la ingeniería de características. Nos centraremos en tres parámetros de la implementación de scikit-learn: n_clusters, max_iter y n_init.

Se trata de un proceso sencillo de dos pasos. El algoritmo comienza inicializando aleatoriamente un número predefinido (n_clusters) de centroides. A continuación, repite estas dos operaciones:

- asignar los puntos al centroide del clúster más cercano
- mover cada centroide para minimizar la distancia a sus puntos

Repite estos dos pasos hasta que los centroides dejen de moverse o hasta que se haya alcanzado un número máximo de iteraciones (max_iter).

A menudo ocurre que la posición aleatoria inicial de los centroides da lugar a una agrupación deficiente. Por este motivo, el algoritmo se repite un número determinado de veces (n_init) y devuelve la agrupación que presenta la menor distancia total entre cada punto y su centroide, es decir, la agrupación óptima.

La animación que aparece a continuación muestra el algoritmo en acción. Ilustra la dependencia del resultado respecto a los centroides iniciales y la importancia de repetir el proceso hasta alcanzar la convergencia.

![imagenes](./images/k-meansClusteringGif.gif)

Es posible que tengas que aumentar el valor de `max_iter` si hay un gran número de clústeres, o el de `n_init` si el conjunto de datos es complejo. No obstante, normalmente el único parámetro que tendrás que elegir tú mismo es `n_clusters` (es decir, `k`). La mejor partición para un conjunto de características depende del modelo que utilices y de lo que intentes predecir, por lo que lo mejor es ajustarla como cualquier hiperparámetro (por ejemplo, mediante validación cruzada).

## **09.4. Analisis de componentes principales**
Se crean dos ejeces de variación los cuales con tangente al gráfico que se ha creado para representar los datos,

![imagenes](./images/analisisDeComponentesPrincipales1.png)

estos ejes se les nombran y despues se hace un desplazamiento del grafico utilizando los nuevos ejes de referencia, y se puede obtener un grafico con los mismos datos pero truncado de cierta forma para que los datos esten representandos de una forma distinta.
![imagene](./images/analisisDeComponentesPrincipales2.png)

Las nuevas características generadas por el análisis de componentes principales (PCA) no son más que combinaciones lineales (sumas ponderadas) de las características originales:

```
df["Size"] = 0.707 * X["Height"] + 0.707 * X["Diameter"]
df["Shape"] = 0.707 * X["Height"] - 0.707 * X["Diameter"]
```

Estas nuevas características se denominan componentes principales de los datos. Las propias ponderaciones se denominan cargas. Habrá tantos componentes principales como características haya en el conjunto de datos original: si hubiéramos utilizado diez características en lugar de dos, habríamos obtenido diez componentes.

### **09.4.1. Analisis de componentes principales para la ingeniria de características**
Hay dos formas de utilizar el Analisis de componentes principales (PCA) para la ingeniería de características.

La primera consiste en utilizarlo como técnica descriptiva. Dado que los componentes proporcionan información sobre la variación, podrías calcular las puntuaciones de MI de los componentes y ver qué tipo de variación es la que mejor predice tu variable objetivo. Esto podría darte ideas sobre qué tipo de características crear: por ejemplo, el producto de ''Altura'' y ''Diámetro'' si el ''Tamaño'' es importante, o la relación entre ''Altura'' y ''Diámetro'' si la ''Forma'' es importante. Incluso podrías intentar realizar un agrupamiento (clustering) en uno o varios de los componentes con puntuaciones más altas.

La segunda forma consiste en utilizar los propios componentes como características. Dado que los componentes revelan directamente la estructura variacional de los datos, a menudo pueden resultar más informativos que las características originales. A continuación se presentan algunos casos de uso:

- Reducción de la dimensionalidad: cuando las características son muy redundantes (concretamente, multicolineales), el PCA separará la redundancia en uno o más componentes de varianza cercana a cero, que luego se pueden descartar, ya que contendrán poca o ninguna información.
- Detección de anomalías: las variaciones inusuales, que no se aprecian en las características originales, suelen aparecer en los componentes de baja varianza. Estos componentes pueden resultar muy informativos en una tarea de detección de anomalías o valores atípicos.
- Reducción del ruido: un conjunto de lecturas de sensores suele compartir cierto ruido de fondo común. El PCA a veces puede concentrar la señal (informativa) en un número menor de características, dejando de lado el ruido, lo que mejora la relación señal-ruido.
- Decorrelación: algunos algoritmos de aprendizaje automático tienen dificultades con características altamente correlacionadas. El PCA transforma las características correlacionadas en componentes no correlacionados, con los que tu algoritmo podría trabajar más fácilmente.

Básicamente, el PCA te proporciona acceso directo a la estructura correlacional de tus datos.

>**Buenas prácticas del PCA**<br>
>Hay algunos aspectos que hay que tener en cuenta al aplicar el PCA:
>
>- El PCA solo funciona con variables numéricas, como cantidades continuas o recuentos.
>- El PCA es sensible a la escala. Es recomendable estandarizar los datos antes de aplicar el PCA, a menos que haya una razón de peso para no hacerlo.
>- Considera la posibilidad de eliminar o restringir los valores atípicos, ya que pueden influir indebidamente en los resultados.

## **09.5. Codificación de destino**
Es cualquier tipo de codificación que sustituye las categorías de una característica por un número derivado del destino.
Este tipo de codificación de la variable objetivo se denomina a veces ''codificación de la media''. Cuando se aplica a una variable objetivo binaria, también se conoce como ''recuento de binos''. (Otros nombres con los que te puedes encontrar son: codificación de verosimilitud, codificación de impacto y codificación ''leave-one-out'').

### **09.5.1. Suavizado**
Sin embargo, una codificación como esta plantea un par de problemas. En primer lugar, están las categorías desconocidas. Las codificaciones de destino conllevan un riesgo especial de sobreajuste, lo que significa que deben entrenarse en una división independiente de ''codificación''. Al unir la codificación a divisiones futuras, Pandas completará los valores que faltan para cualquier categoría que no esté presente en la división de codificación. Estos valores que faltan habría que imputarlos de alguna manera.

En segundo lugar, están las categorías poco frecuentes. Cuando una categoría solo aparece unas pocas veces en el conjunto de datos, es poco probable que las estadísticas calculadas sobre ese grupo sean muy precisas. En el conjunto de datos ''Automobiles'', la marca ''Mercury'' solo aparece una vez. El precio ''medio'' que hemos calculado es simplemente el precio de ese único vehículo, lo que podría no ser muy representativo de los Mercury que podamos encontrar en el futuro. La codificación de destino de categorías poco frecuentes puede aumentar la probabilidad de sobreajuste.

Una solución a estos problemas es aplicar un suavizado. La idea consiste en combinar la media dentro de la categoría con la media general. Las categorías poco frecuentes reciben menos peso en su media de categoría, mientras que las categorías con valores faltantes simplemente reciben la media general.

En pseudocódigo:
```
codificación = peso * media_dentro_de_la_categoría + (1 - peso) * media_general
```

donde peso es un valor entre 0 y 1 calculado a partir de la frecuencia de la categoría.

donde ''peso'' es un valor comprendido entre 0 y 1 que se calcula a partir de la frecuencia de la categoría.

Una forma sencilla de determinar el valor de ''peso'' es calcular una estimación m:

```
peso = n / (n + m)
```

donde n es el número total de veces que aparece esa categoría en los datos. El parámetro m determina el ''factor de suavizado''. Cuanto mayor sea el valor de m, mayor será el peso que se le da a la estimación global.

![imagen](./images/suavizadoCodificacionObjetivo.png)

> **Casos de uso de la codificación de objetivo**<br>
> La codificación de objetivo resulta ideal para:
> 
>- **Características de alta cardinalidad**: una característica con un gran número de categorías puede resultar complicada de codificar: una codificación ''one-hot'' generaría demasiadas características y otras alternativas, como la codificación de etiquetas, podrían no ser adecuadas para esa característica. La codificación de objetivo deriva valores numéricos para las categorías utilizando la propiedad más importante de la característica: su relación con el objetivo.
>- **Características motivadas por el dominio**: por experiencia previa, es posible que se sospeche que una característica categórica debería ser importante aunque haya obtenido una puntuación baja en una métrica de características. La codificación de destino puede ayudar a revelar el verdadero valor informativo de una característica.

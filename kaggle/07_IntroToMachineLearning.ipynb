## **07.2. Mi primer modelo de Machine Learning**

### **07.2.1. Seleccionar datos para el modelaje**
- `db.columns`: muestra solo las columnas.

````python
import pandas as pd

melbourne_file_path = './melb_data.csv'
melbourne_data = pd.read_csv(melbourne_file_path) 
melbourne_data.columns
````

Se eliminan los valores nulos o que no se han introducido para poder entrenar el modelo.

````python
# dropna drops missing values (think of na as "not available")
melbourne_data = melbourne_data.dropna(axis=0)
````

#### **07.2.1.1. Seleccionar un objetivo de prediccion**
Puedes extraer una variable mediante la notación de puntos. Esta única columna se almacena en una ''Series'', que, a grandes rasgos, es como un ''DataFrame'' con una sola columna de datos.

Utilizaremos la notación de puntos para seleccionar la columna que queremos predecir, lo que se denomina ''objetivo de predicción''. Por convención, el objetivo de predicción se denomina ''y''. Así pues, el código que necesitamos para guardar los precios de las viviendas en los datos de Melbourne es

````python
y = melbourne_data.Price
````

### **07.2.2. Seleccionar ''caracteristicas''**
Las columnas que se introducen en nuestro modelo (y que posteriormente se utilizan para realizar predicciones) se denominan ''características''. En nuestro caso, serían las columnas utilizadas para determinar el precio de la vivienda. A veces, se utilizarán como características todas las columnas excepto la variable objetivo. Otras veces, será mejor utilizar menos características.

Por ahora, construiremos un modelo con solo unas pocas características. Más adelante verás cómo iterar y comparar modelos construidos con diferentes características.

Seleccionamos varias características indicando una lista de nombres de columnas entre corchetes. Cada elemento de esa lista debe ser una cadena de caracteres (entre comillas).

````python
melbourne_features = ['Rooms', 'Bathroom', 'Landsize', 'Lattitude', 'Longtitude']
````

Por convención, estos datos se denominan *X.f*.

````python
X = melbourne_data[melbourne_features]
````

Revisión antes de entrenar el modelo de los datos que se le van a introducir.

````python
X.describe()
````

- `head()`: permite ver las cinco primeras filas del archivo csv.

````python
X.head()
````

### **07.2.3. Construir el modelo**
Se utilizara la biblioteca *scikit-learn* para crear tus modelos. Al programar, esta biblioteca se escribe como ''sklearn'', tal y como verás en el código de ejemplo. Scikit-learn es, sin duda, la biblioteca más popular para modelar los tipos de datos que suelen almacenarse en DataFrames.

Los pasos para crear y utilizar un modelo son los siguientes:

- *Definir*: ¿Qué tipo de modelo será? ¿Un árbol de decisión? ¿Algún otro tipo de modelo? También se especifican otros parámetros del tipo de modelo.
- *Ajuste*: Capturar patrones a partir de los datos proporcionados. Esta es la esencia del modelado.
- *Predicción*: Tal y como suena.
- *Evaluación*: Determinar la precisión de las predicciones del modelo.

A continuación se muestra un ejemplo de cómo definir un modelo de árbol de decisión con scikit-learn y ajustarlo con las características y la variable objetivo.

---

- *fit()*: Divide los datos de entrenamiento del modelo en dos.

````python
from sklearn.tree import DecisionTreeRegressor

# Define model. Specify a number for random_state to ensure same results each run
melbourne_model = DecisionTreeRegressor(random_state=1)

# Fit model
melbourne_model.fit(X, y)
````

Muchos modelos de aprendizaje automático admiten cierto grado de aleatoriedad en el entrenamiento del modelo. Al especificar un número para `random_state`, te aseguras de obtener los mismos resultados en cada ejecución. Esto se considera una buena práctica. Puedes utilizar cualquier número, y la calidad del modelo no dependerá de manera significativa del valor concreto que elijas.

Ahora disponemos de un modelo ajustado que podemos utilizar para realizar predicciones.

En la práctica, lo más habitual es realizar predicciones sobre las nuevas viviendas que salen al mercado, en lugar de sobre aquellas de las que ya disponemos de precios. No obstante, realizaremos predicciones para las primeras filas de los datos de entrenamiento con el fin de comprobar cómo funciona la función `predict`.


````python
print("Making predictions for the following 5 houses:")
print(X.head())
print("The predictions are")
print(melbourne_model.predict(X.head()))
#print(melbourne_model.predict(X))
````

## **07.3. Validacion del modelo**

### **07.3.1. ¿Que es la validación de modelos**
Interesa evaluar casi todos los modelos que construyas. En la mayoría de las aplicaciones (aunque no en todas), la medida relevante de la calidad de un modelo es la precisión predictiva. En otras palabras, si las predicciones del modelo se acercarán a lo que ocurre realmente.

Muchas personas cometen un grave error al medir la precisión predictiva. Realizan predicciones con sus datos de entrenamiento y comparan esas predicciones con los valores objetivo de los propios datos de entrenamiento. Enseguida verás cuál es el problema de este enfoque y cómo resolverlo, pero primero pensemos en cómo lo haríamos.

En primer lugar, tendrías que resumir la calidad del modelo de una forma comprensible. Si comparas los valores predichos y reales de 10 000 viviendas, probablemente encontrarás una mezcla de predicciones acertadas y erróneas. Revisar una lista de 10 000 valores predichos y reales no tendría sentido. Necesitamos resumir esto en una única métrica.

Existen muchas métricas para resumir la calidad del modelo, pero empezaremos con una llamada ''error absoluto medio'' (también conocido como MAE). Analicemos esta métrica empezando por la última palabra: ''error''.

El error de predicción para cada vivienda es:

```
error = valor real − valor previsto
```

Así pues, si una vivienda cuesta `150 000 $` y has previsto que costaría `100 000 $`, el error es de `50 000 $`.

Con la métrica MAE, tomamos el valor absoluto de cada error. Esto convierte cada error en un número positivo. A continuación, calculamos la media de esos errores absolutos. Esta es nuestra medida de la calidad del modelo. En lenguaje sencillo, se puede expresar así:

```
De media, nuestras predicciones tienen un error de aproximadamente X.
```

Para calcular el MAE, primero necesitamos un modelo. Este se ha creado en una celda oculta más abajo, que puedes consultar haciendo clic en el botón de código.

````python
from sklearn.metrics import mean_absolute_error

predicted_home_prices = melbourne_model.predict(X)
mean_absolute_error(y, predicted_home_prices)
````

### **07.3.2. El problema con los resultados de "In-Sample"**
La medida que acabamos de calcular puede denominarse ''puntuación dentro de la muestra''. Hemos utilizado una única ''muestra'' de viviendas tanto para construir el modelo como para evaluarlo. A continuación explicamos por qué esto es un problema.

Imagina que, en el amplio mercado inmobiliario, el color de la puerta no guarda relación alguna con el precio de la vivienda.

Sin embargo, en la muestra de datos que has utilizado para construir el modelo, todas las viviendas con puertas verdes eran muy caras. La función del modelo es encontrar patrones que predigan los precios de las viviendas, por lo que detectará este patrón y siempre predecirá precios elevados para las viviendas con puertas verdes.

Dado que este patrón se ha extraído de los datos de entrenamiento, el modelo parecerá preciso en dichos datos.

Pero si este patrón no se cumple cuando el modelo analiza datos nuevos, el modelo resultaría muy impreciso al aplicarlo en la práctica.

Dado que el valor práctico de los modelos radica en realizar predicciones sobre datos nuevos, medimos su rendimiento con datos que no se utilizaron para construir el modelo. La forma más sencilla de hacerlo es excluir algunos datos del proceso de construcción del modelo y, a continuación, utilizarlos para comprobar la precisión del modelo con datos que no ha visto antes. Estos datos se denominan datos de validación.

### **07.3.3. Como programarlo**
La biblioteca scikit-learn cuenta con una función llamada `train_test_split` que permite dividir los datos en dos partes. Utilizaremos una parte de esos datos como datos de entrenamiento para ajustar el modelo, y la otra parte como datos de validación para calcular el `mean_absolute_error`.

````python
from sklearn.model_selection import train_test_split

# split data into training and validation data, for both features and target
# The split is based on a random number generator. Supplying a numeric value to
# the random_state argument guarantees we get the same split every time we
# run this script.
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state = 0)
# Define model
melbourne_model = DecisionTreeRegressor()
# Fit model
melbourne_model.fit(train_X, train_y)

# get predicted prices on validation data
val_predictions = melbourne_model.predict(val_X)
print(mean_absolute_error(val_y, val_predictions))
print("Exito!")
````

Tu error absoluto medio para los datos de la muestra fue de unos 500 dólares. Fuera de la muestra, supera los 250 000 dólares.

Esta es la diferencia entre un modelo que es casi totalmente preciso y otro que resulta inutilizable para la mayoría de los fines prácticos. A modo de referencia, el valor medio de una vivienda en los datos de validación es de 1,1 millones de dólares. Por lo tanto, el error en los nuevos datos es aproximadamente una cuarta parte del valor medio de una vivienda.

Hay muchas formas de mejorar este modelo, como experimentar para encontrar mejores características o diferentes tipos de modelos.

## **08.1. Valores desaparecidos o no introducidos**
Hay tres formas de trabajar con valores desaparecidos o que no se han introducido las cuales son:
- <ins>*Eliminar las columnas con los valores desaperidos o que no se han introducido*</ins>: Es la opción más simple la cual simplemente elimina la columna cuando uno de los valores es NaN (Not a Number), lo malo de este método es que el modelo pierde acceso a mucha información que puede ser interesante para procesarla, ya que si un unico valor es NaN en una columna esta se desecha entera.
![imagen](./images/eliminarLasColumnasConValoresDesaparecidosTabla.png)
- <ins>*Atribución de valores donde hay celdas con valores NaN*</ins>: La atribución rellena los valores no introducidos o desaparecidos con algun tipo de número. Por ejemplo se puede rellenar dichos valores no introducidos con el valor de la columna contigua, el valor nuevo en muchos de los casos no estara exactamente bien pero normalmente los modelos son mas precisos con esta forma que eliminando la columna entera.
![imagen](./images/atribucionDeValoresDondeHayCeldasConValoresNaNTabla.png)
- <ins>*Extension de la atribución de valores donde hay celdas con valores NaN*</ins>: La atribución de valores normalmente funciona bien y es la forma de realizarlo en primera instancia, de todas formas los valores que se han introducido por copia de la columna adyacente, y por lo tanto los valores introducidos pueden estar sistematicamente por encima o por debajo de los valores que en verdad representan, o dichos valores son unicos en cierta forma ya que pueden significar algo distinto que el valor atribuido, por lo tanto esta nueva metodologia lo que implementa es el metodo anterior por lo tanto el despligue es el mismo pero se incluye una columna nueva que con valores booleanos o entendibles para el modelo para ver si dicho valores ettaba desaparecido, en algunos casos los modelos funcionan mejor y en otros casos no hay diferencia con el anterior punto.

## **08.2. Variables categorizadas**
Las variables categorizadas toman un numero limitado de opciones, por lo tanto refiriendose a por ejemplo cuando se responde una encuesta por internet, esto es malo porque algunos modelos no las aceptan, por ejemplo te pueden dar cuatro valores para responder "Nunca", "Raramente", "Casi todos los dias" o "Todos los dias". Por lo tanto para tratar con este tipo de variables existen tres métodos distintos los cuales son:
- <ins>*Eliminar variables categorizadas*</ins>: Lo cual significa que se desechan las variables del dataset las cuales tengan este tipo de valores, pero como siempre con este tipo de metodos se pierden valores que pueden ser importantes para el modelo.
- <ins>*Encodificación ordinal*</ins>: Asigna los valores catagorizados a un entero, por lo tanto una opción es igual a 0, otro es igual a 1 y viceversa. Esta suposición tiene sentido en este ejemplo, ya que existe una jerarquía indiscutible entre las categorías. No todas las variables categóricas presentan un orden claro en sus valores, pero a aquellas que sí lo tienen las denominamos variables ordinales. En el caso de los modelos basados en árboles (como los árboles de decisión y los bosques aleatorios), cabe esperar que la codificación ordinal funcione bien con las variables ordinales.
![image](./images/enodificacionOrdinalTable.png)
- <ins>*Encodificacion One-Hot*</ins>: La codificación ''one-hot'' crea nuevas columnas que indican la presencia (o ausencia) de cada valor posible en los datos originales. En el conjunto de datos original, ''Color'' es una variable categórica con tres categorías: ''Rojo'', ''Amarillo'' y ''Verde''. La codificación ''one-hot'' correspondiente contiene una columna para cada valor posible y una fila para cada fila del conjunto de datos original. Siempre que el valor original fuera ''Rojo'', ponemos un 1 en la columna ''Rojo''; si el valor original fuera ''Amarillo'', ponemos un 1 en la columna ''Amarillo'', y así sucesivamente. A diferencia de la codificación ordinal, la codificación ''one-hot'' no supone un orden entre las categorías. Por lo tanto, cabe esperar que este enfoque funcione especialmente bien si no existe un orden claro en los datos categóricos (por ejemplo, ''Rojo'' no es ni más ni menos que ''Amarillo''). Nos referimos a las variables categóricas sin una jerarquía intrínseca como variables nominales. La codificación ''one-hot'' no suele dar buenos resultados si la variable categórica adopta un gran número de valores (es decir, por lo general no se utiliza para variables que adopten más de 15 valores diferentes).

## **08.3. Tuberias(Pipelines)**
Los flujos de trabajo son una forma sencilla de mantener organizado el código de preprocesamiento y modelización de datos. En concreto, un flujo de trabajo agrupa los pasos de preprocesamiento y modelización para que puedas utilizar todo el conjunto como si fuera un único paso.

Muchos científicos de datos crean modelos de forma improvisada sin utilizar pipelines, pero estos ofrecen algunas ventajas importantes. Entre ellas se incluyen:

- <ins>*Código más limpio*</ins>: llevar la cuenta de los datos en cada paso del preprocesamiento puede resultar complicado. Con un pipeline, no tendrás que realizar un seguimiento manual de tus datos de entrenamiento y validación en cada paso.
- <ins>*Menos errores*</ins>: hay menos posibilidades de aplicar mal un paso u olvidarse de un paso del preprocesamiento.
- <ins>*Mayor facilidad para la puesta en producción*</ins>: puede resultar sorprendentemente difícil pasar de un prototipo a un modelo que se pueda implementar a gran escala. No vamos a entrar aquí en las numerosas cuestiones relacionadas, pero los pipelines pueden ayudar.
- <ins>*Más opciones para la validación de modelos*</ins>: verás un ejemplo en el siguiente tutorial, que trata soinse la validación cruzada.


## **08.4. Validación cruzada (Cross-Validation)**
En la validación cruzada, aplicamos nuestro proceso de modelización a diferentes subconjuntos de los datos para obtener múltiples medidas de la calidad del modelo.

Por ejemplo, podríamos empezar dividiendo los datos en 5 partes, cada una de las cuales represente el 20 % del conjunto de datos completo. En este caso, decimos que hemos dividido los datos en 5 «pliegues».

![images](./images/cross-validationTable.png)

A continuación, realizamos un experimento para cada pliegue:

- En el Experimento 1, utilizamos el primer pliegue como conjunto de validación (o de retención) y el resto como datos de entrenamiento. Esto nos proporciona una medida de la calidad del modelo basada en un conjunto de retención del 20 %.
- En el Experimento 2, reservamos los datos del segundo pliegue (y utilizamos todo excepto el segundo pliegue para entrenar el modelo). A continuación, el conjunto de reserva se utiliza para obtener una segunda estimación de la calidad del modelo.
- Repetimos este proceso, utilizando cada pliegue una vez como conjunto de validación. En resumen, el 100 % de los datos se utiliza como conjunto de validación en algún momento, y obtenemos una medida de la calidad del modelo basada en todas las filas del conjunto de datos (aunque no utilicemos todas las filas simultáneamente).


### **08.4.1. ¿Cuando hay que usar la validación cruzada**
La validación cruzada ofrece una medida más precisa de la calidad del modelo, lo cual es especialmente importante si tienes que tomar muchas decisiones relacionadas con la modelización. Sin embargo, su ejecución puede llevar más tiempo, ya que estima varios modelos (uno por cada pliegue).

Así pues, teniendo en cuenta estas ventajas e inconvenientes, ¿cuándo se debe utilizar cada enfoque?

- Para conjuntos de datos pequeños, en los que la carga computacional adicional no supone un gran problema, se debe utilizar la validación cruzada.
- Para conjuntos de datos más grandes, basta con un único conjunto de validación. El código se ejecutará más rápido y es posible que se disponga de datos suficientes como para que apenas sea necesario reutilizar parte de ellos para el conjunto de prueba.

No existe un umbral sencillo para distinguir entre un conjunto de datos grande y uno pequeño. Pero si tu modelo tarda un par de minutos o menos en ejecutarse, probablemente merezca la pena pasar a la validación cruzada.

Como alternativa, puedes ejecutar la validación cruzada y comprobar si las puntuaciones de cada experimento parecen similares. Si cada experimento arroja los mismos resultados, probablemente baste con un único conjunto de validación.

## **08.5. Mejora por gradiente (XGBoost)**
El ''gradient boosting'' es un método que sigue una serie de ciclos para ir añadiendo modelos de forma iterativa a un conjunto.

Comienza inicializando el conjunto con un único modelo, cuyas predicciones pueden ser bastante rudimentarias. (Aunque sus predicciones sean tremendamente inexactas, las incorporaciones posteriores al conjunto corregirán esos errores).

A continuación, iniciamos el ciclo:

- En primer lugar, utilizamos el conjunto actual para generar predicciones para cada observación del conjunto de datos. Para realizar una predicción, sumamos las predicciones de todos los modelos del conjunto.
- Estas predicciones se utilizan para calcular una función de pérdida (como el error cuadrático medio, por ejemplo).
- A continuación, utilizamos la función de pérdida para ajustar un nuevo modelo que se añadirá al conjunto. Concretamente, determinamos los parámetros del modelo de tal forma que añadir este nuevo modelo al conjunto reduzca la pérdida. (Nota al margen: el término «gradiente» en «gradient boosting» hace referencia al hecho de que utilizaremos el descenso por gradiente sobre la función de pérdida para determinar los parámetros de este nuevo modelo.)
- Por último, añadimos el nuevo modelo al conjunto y...
- ... ¡repetimos!
![imagen](./images/gradientBoostTabla.png)

### **08.5.1. Ajustes de parametros**\
XGBoost cuenta con algunos parámetros que pueden afectar de forma significativa a la precisión y a la velocidad de entrenamiento. Los primeros parámetros que debes comprender son:
n_estimators

n_estimators especifica cuántas veces se debe repetir el ciclo de modelización descrito anteriormente. Equivale al número de modelos que incluimos en el conjunto.

- Un valor demasiado bajo provoca un ajuste insuficiente, lo que da lugar a predicciones inexactas tanto en los datos de entrenamiento como en los de prueba.
- Un valor demasiado alto provoca un sobreajuste, lo que da lugar a predicciones precisas en los datos de entrenamiento, pero imprecisas en los de prueba (que es lo que nos interesa).

Los valores habituales oscilan entre 100 y 1000, aunque esto depende en gran medida del parámetro learning_rate que se explica más adelante.

## **08.6. Fuga de datos**
La fuga de datos (o ''leakage'') se produce cuando los datos de entrenamiento contienen información sobre la clase objetivo, pero no se dispone de datos similares cuando el modelo se utiliza para la predicción. Esto da lugar a un alto rendimiento en el conjunto de entrenamiento (y posiblemente incluso en los datos de validación), pero el modelo tendrá un rendimiento deficiente en producción.

En otras palabras, la fuga hace que un modelo parezca preciso hasta que se empiezan a tomar decisiones con él, momento en el que el modelo se vuelve muy impreciso.

Existen dos tipos principales de fuga: la fuga de la clase objetivo y la contaminación entre el conjunto de entrenamiento y el de prueba.

### **08.6.1. Fuga del objetivo (Target leakage)**
La fuga de datos en el aprendizaje automático se produce cuando un modelo utiliza durante el entrenamiento información que no estaría disponible en el momento de la predicción. La fuga hace que un modelo predictivo parezca preciso hasta que se implementa en su caso de uso; a partir de ese momento, ofrecerá resultados inexactos, lo que dará lugar a una toma de decisiones deficiente y a conclusiones erróneas.

Para prevenir este tipo de fuga de datos, cualquier variable actualizada o creada depues de que el valor que se quiere obtener en validación deberia ser excluida.

### **08.6.2. Contaminación en la prueba de entrenamiento (Train-Test Contamination)**
Se produce otro tipo de fuga cuando no se tiene cuidado a la hora de distinguir los datos de entrenamiento de los de validación.

Recuerda que la validación está pensada para medir el rendimiento del modelo con datos que no ha tenido en cuenta anteriormente. Este proceso puede verse alterado de forma sutil si los datos de validación influyen en el comportamiento del preprocesamiento. A esto se le denomina a veces ''contaminación entre entrenamiento y prueba''.

Por ejemplo, imagina que realizas un preprocesamiento (como ajustar un imputador para los valores perdidos) antes de llamar a train_test_split(). ¿Cuál es el resultado final? Tu modelo puede obtener buenas puntuaciones de validación, lo que te da mucha confianza en él, pero puede funcionar mal cuando lo implementes para tomar decisiones.

Al fin y al cabo, has incorporado datos de la validación o de la prueba en tu forma de realizar predicciones, por lo que el modelo puede funcionar bien con esos datos concretos, aunque no sea capaz de generalizar a datos nuevos. Este problema se vuelve aún más sutil (y más peligroso) cuando se lleva a cabo una ingeniería de características más compleja.

Si tu validación se basa en una simple división entre entrenamiento y prueba, excluye los datos de validación de cualquier tipo de ajuste, incluido el ajuste de los pasos de preprocesamiento. Esto resulta más sencillo si utilizas los pipelines de scikit-learn. Cuando utilices la validación cruzada, ¡es aún más importante que realices el preprocesamiento dentro del pipeline!

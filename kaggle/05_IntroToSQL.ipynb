## **05.1. Empezando con SQL y Pandas**

### **05.1.1. Introducción y preparación del entorno**

Abre el archivo sql local y permite hacer preguntas a la base de datos, esta siguiente es para obtener todas las tablas dentro de la base de datos.

```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("sql-murder-mystery.db")

query = """
SELECT name
FROM sqlite_master
WHERE type='table';
"""

# Consultar el catálogo del sistema para listar las tablas
tablas = pd.read_sql_query(query, conexion)
display(tablas)

# Cerrar la conexión al terminar
conexion.close()
```

### **05.1.2. Obtención del esquema de creación de la tabla**

```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("sql-murder-mystery.db")

# Realizar tu consulta SQL
query = """
SELECT sql
FROM sqlite_master
WHERE type='table'
AND name='crime_scene_report';
"""

db = pd.read_sql_query(query, conexion)

# Imprimir por pantalla con iloc[] todo el contenido de la celda sql
print(db["sql"].iloc[0])

# Cerrar la conexión al terminar
conexion.close()
```

## **05.2. SELECT, FROM & WHERE**

### **05.2.1. SELECT... FROM**
La consulta SQL más básica selecciona una sola columna de una sola tabla. Para ello, indica la columna que desees después de la palabra SELECT y, a continuación, indica la tabla después de la palabra FROM.

### **05.2.2. WHERE...**
La función WHERE dentro de SQL sirve para filtrar la la respuesta de la pregunta que le hacemos a la base de datos, de forma que nos devuelva menos datos.

```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("sql-murder-mystery.db")

# Realizar tu consulta SQL
query = """
SELECT *
FROM 'crime_scene_report'
WHERE type='murder';
"""

db = pd.read_sql_query(query, conexion)

display(db.loc[:10,:])

# Cerrar la conexión al terminar
conexion.close()
```

## **05.3. GROUP BY, HAVING & COUNT**

### **05.3.1. COUNT()**
`COUNT()` como su nombre indica sirve para contar cuantas veces aparece un valor en la vista de la base de datos, por ejemplo en el siguiente código se cuentan cuantas ciudades dentro de la tabla de ''crime_scene_report'' tienen el tipo de ''murder''.

```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("sql-murder-mystery.db")

# Realizar tu consulta SQL
query = """
SELECT COUNT(city)
FROM 'crime_scene_report'
WHERE type='murder';
"""

db = pd.read_sql_query(query, conexion)

display(db.loc[:10,:])

# Cerrar la conexión al terminar
conexion.close()
```

### **05.3.2. GROUP BY**
La instrucción GROUP BY toma el nombre de una o varias columnas y trata todas las filas que tengan el mismo valor en esa columna como un único grupo cuando se aplican funciones de agregación como COUNT().

```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("sql-murder-mystery.db")

# Realizar tu consulta SQL
query = """
SELECT city, COUNT(*) AS total_homicidios
FROM 'crime_scene_report'
WHERE type='murder'
GROUP BY city;
"""

db = pd.read_sql_query(query, conexion)

display(db.loc[:,:])

# Cerrar la conexión al terminar
conexion.close()
```

### **05.3.3. GROUP BY... HAVING**
La cláusula HAVING se utiliza junto con GROUP BY para ignorar los grupos que no cumplen determinados criterios.

```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("sql-murder-mystery.db")

# Realizar tu consulta SQL
query = """
SELECT city, COUNT(*) AS total_homicidios
FROM 'crime_scene_report'
WHERE type='murder'
GROUP BY city
HAVING COUNT(*) > 1;
"""

# Tambien se puede hacer asi
#query = """
#SELECT city, COUNT(*) AS total_homicidios
#FROM 'crime_scene_report'
#WHERE type='murder'
#GROUP BY city
#HAVING total_homicidios > 1;
#"""

db = pd.read_sql_query(query, conexion)

display(db.loc[:,:])

# Cerrar la conexión al terminar
conexion.close()
```

## **05.5. AS & WITH**

### **05.5.1. AS**
En un tutorial anterior aprendiste a utilizar ''AS'' para renombrar las columnas generadas por tus consultas, lo que también se conoce como ''alias''. Esto es similar a cómo Python utiliza ''as'' para crear alias al realizar importaciones como ''import pandas as pd'' o ''import seaborn as sns''.

```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("Chinook_Sqlite.sqlite")

# Realizar tu consulta SQL
query = """
SELECT City, COUNT(SupportRepId) AS TotalClientes
FROM 'Customer'
GROUP BY City
ORDER BY TotalClientes DESC;
"""

db = pd.read_sql_query(query, conexion)

display(db.loc[:10,:])

# Cerrar la conexión al terminar
conexion.close()
```

### **05.5.2. WITH... AS**
Por sí sola, la cláusula `AS` es una forma práctica de depurar los datos devueltos por la consulta. Resulta aún más potente cuando se combina con la cláusula `WITH` en lo que se denomina una ''expresión de tabla común''.

Una expresión de tabla común (o CTE) es una tabla temporal que se devuelve dentro de la consulta. Las CTE son útiles para dividir las consultas en fragmentos más legibles, y se pueden escribir consultas sobre ellas.

```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("Chinook_Sqlite.sqlite")

# Realizar tu consulta SQL
query = """
WITH Something AS
(
    SELECT City, COUNT(SupportRepId) AS TotalClientes
    FROM 'Customer'
    GROUP BY City
    ORDER BY TotalClientes DESC
)
SELECT totalClientes
FROM Something;
"""

db = pd.read_sql_query(query, conexion)

display(db.loc[:10,:])

# Cerrar la conexión al terminar
conexion.close()
```

## **05.6. Joining Data**

### **05.6.1. JOIN**

```python
import sqlite3
import pandas as pd

# Conectar al archivo
conexion = sqlite3.connect("Chinook_Sqlite.sqlite")

# Realizar tu consulta SQL
query = """
SELECT ar.Name, al.Title
FROM Artist AS ar
INNER JOIN Album AS al
ON ar.ArtistId = al.ArtistID;
"""

db = pd.read_sql_query(query, conexion)

display(db.loc[:10,:])

# Cerrar la conexión al terminar
conexion.close()
```

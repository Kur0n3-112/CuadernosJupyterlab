#### **02.2.2. Indexacion en pandas**

Esto significa que resulta ligeramente más fácil recuperar filas y ligeramente más difícil recuperar columnas.

##### **02.2.2.3. Elegir entre `loc` e `iloc`**
A la hora de elegir o cambiar entre `loc` e `iloc`, hay un «matiz» que conviene tener en cuenta: ambos métodos utilizan esquemas de indexación ligeramente diferentes.

`iloc` utiliza el esquema de indexación de la biblioteca estándar de Python, en el que se incluye el primer elemento del rango y se excluye el último. Por lo tanto, 0:10 seleccionará las entradas 0, ..., 9. loc, por su parte, indexa de forma inclusiva. Así, 0:10 seleccionará las entradas 0, ..., 10.

¿A qué se debe este cambio? Recuerda que loc puede indexar cualquier tipo de la biblioteca estándar: cadenas, por ejemplo. Si tenemos un DataFrame con valores de índice Apples, ..., «Potatoes», …, y queremos seleccionar «todas las frutas en orden alfabético entre Apples y Potatoes», resulta mucho más cómodo indexar df.loc[“Apples”:'Potatoes'] que indexar algo como df.loc[“Apples”, “Potatoet”] (la «t» viene después de la «s» en el alfabeto).

Esto resulta especialmente confuso cuando el índice del DataFrame es una simple lista numérica, por ejemplo, 0,...,1000. En este caso, df.iloc[0:1000] devolverá 1000 entradas, mientras que df.loc[0:1000] devolverá 1001. Para obtener 1000 elementos utilizando `loc`, tendrás que bajar un valor y solicitar `df.loc[0:999]`.

Por lo demás, la semántica del uso de `loc` es la misma que la de `iloc`.

### **02.3.2. Mapas**
El término «mapa», tomado de las matemáticas, se refiere a una función que toma un conjunto de valores y los «asigna» a otro conjunto de valores. En la ciencia de datos, a menudo necesitamos crear nuevas representaciones a partir de datos existentes, o transformar los datos del formato en el que se encuentran ahora al formato que queremos que tengan más adelante. Los mapas son los que se encargan de esta tarea, ¡lo que los convierte en elementos extremadamente importantes para llevar a cabo tu trabajo!

- `(lambda p :)`-> Define una funcion rapida e innominada la cual recibe como parametro de entrada p el cual es un valor individual o un elemento de la serie.

- `(p - review_points_mean)`-> Le resta dicho valor p a la variable `review_points_mean`.

```python
review_points_mean = reviews.points.mean()
reviews.points.map(lambda p: p - review_points_mean)
```

- Devuelve una tabla la cual tien alterada los valores de la columna "points".
- Se puede cambiar el valor de `axis` para que en vez de ir por columnas vaya por filas (`axis=index`).
- `apply` conmuta cambios en la tabla segun los valores que se le pasen.

```python
def remean_points(row):
    row.points = row.points - review_points_mean
    return row

reviews.apply(remean_points, axis='columns')
```

- Realiza lo mismo que en anterior apartado pero con las operaciones integradas en Pandas, por lo tanto estas contienen unos metodos mas optimizados y por ende son mas rapidos y eficientes.

```python
review_points_mean = reviews.points.mean()
reviews.points - review_points_mean
```

- En este código estamos realizando una operación entre muchos valores en el lado izquierdo (todo lo que hay en la serie) y un único valor en el lado derecho (el valor medio). Pandas analiza esta expresión y deduce que lo que queremos hacer es restar ese valor medio de cada valor del conjunto de datos.

- Pandas también sabrá qué hacer si realizamos estas operaciones entre Series de la misma longitud. Por ejemplo, una forma sencilla de combinar la información de país y región en el conjunto de datos sería hacer lo siguiente.

```python
reviews.country + " - " + reviews.region_1
```

Estos operadores son más rápidos que map() o apply(), ya que utilizan las optimizaciones integradas en pandas. Todos los operadores estándar de Python (>, <, ==, etc.) funcionan de esta manera.

Sin embargo, no son tan flexibles como map() o apply(), que permiten realizar operaciones más avanzadas, como aplicar lógica condicional, algo que no se puede hacer solo con la suma y la resta.

## **02.4. Agrupaciones y Filtrado**

### 02.4.1. Analisis por grupos


- `groupby()`: funcion la cual agrupa los valores almacenados en las columnas que se le pasen como argumento a esta funcion.
- `count()`: funcion del grupo de funciones *groupby()* la cual permite contar los valores agrupados.

```python
reviews.groupby('points').points.count()
```

- `size()`: en este contexto hace lo mismo que *count()* pero es mas escueto en algunos aspectos.

```python
reviews.groupby('points').size()
```

- `min()`: funcion del grupo de funciones de *groupby()* la cual sirve para devolver los minimos dentro de un grupo.
- `reviews.groupby('points').price.min()`: Se entiende como agrupar por el valor de los puntos, el valor de la columna de precios, y de esa columna de precios devuelve el minimo valor que se asocie a la columna de puntos.

>Interpretacion de la IA para mas ayuda al entendimiento de esta parte:
>* **`reviews.groupby('points')`**: Separa la estructura de datos en grupos según los valores únicos de la columna **`points`**.
>* **`.price`**: Selecciona únicamente la columna de precios para cada uno de los grupos formados.
>* **`.min()`**: Calcula el valor más bajo dentro de la columna de precios para cada grupo.
>
> El resultado es una Serie de Pandas donde el índice son los puntajes de la columna **`points`** y sus valores asociados son los precios mínimos encontrados para cada puntaje.

```python
reviews.groupby('points').price.min()
```

Podemos considerar que cada grupo que generamos es una sección de nuestro DataFrame que contiene únicamente datos con valores coincidentes. Podemos acceder directamente a este DataFrame mediante el método *apply()*, y a continuación podemos manipular los datos como mejor nos parezca. Por ejemplo, esta es una forma de seleccionar el nombre del primer vino reseñado de cada bodega del conjunto de datos:

```python
reviews.groupby('winery').apply(lambda df: df.title.iloc[0])
```

Se puede tener un control mas precioso, tambien se puede agrupar por mas de una columna, un ejemplo de esto esta en el siguiente bloque de codigo.

```python
reviews.groupby(['country', 'province']).apply(lambda df: df.loc[df.points.idxmax()])
```

- `agg()`: pertenece al grupo de funciones de *groupby()* esta en especifico permite ejecutar diferentes funciones en el DataFrame de forma simultanea.

```python
reviews.groupby(['country']).price.agg([len, "min", "max"])
```

### **02.4.2. Indices multiples**

Los indices multiples se diferencian de los indices normales porque estos tienen varios niveles.

```python
countries_reviewed = reviews.groupby(['country', 'province']).description.agg([len])
countries_reviewed
```

```python
mi = countries_reviewed.index
type(mi)
```

Los índices múltiples cuentan con varios métodos para gestionar su estructura por niveles, que no existen en los índices de un solo nivel. Además, requieren dos niveles de etiquetas para recuperar un valor. Gestionar los resultados de los índices múltiples suele ser una «trampa» habitual para los usuarios que se inician en pandas.

Los casos de uso de un índice múltiple se detallan junto con las instrucciones para utilizarlos en la sección «MultiIndex / Advanced Selection» de la documentación de pandas.

Sin embargo, en general, el método de índice múltiple que utilizarás con más frecuencia es el que sirve para volver a convertirlo en un índice normal, el método `reset_index()`.


```python
countries_reviewed.reset_index()
```

### **02.4.3. Filtrado**

Usando la tabla que se genero en el indice multiple anterior, como se puede comprobar los valores de dicha tabla no estaban ordenados, por lo tanto se pued usar *sort_values()* para realizar esta ordenacion.

- `sort_values()`: funcion del grupo de funciones *groupby()* la cual permite ordenar la tabla por un valor que se le adjunta como clave del orden.

```python
countries_reviewed = countries_reviewed.reset_index()
countries_reviewed.sort_values(by='len')
```

La función *sort_values()* ordena por defecto de forma ascendente, es decir, los valores más bajos aparecen primero. Sin embargo, la mayoría de las veces queremos una ordenación descendente, en la que los números más altos aparezcan primero.

```python
countries_reviewed.sort_values(by='len', ascending=False)
```

Se puede filtrar mas de una columna a la vez.

```python
countries_reviewed.sort_values(by=['country', 'len'])
```

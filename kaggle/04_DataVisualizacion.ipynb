## **04.3. Gráficos de barras y mapas de calor**

### **04.3.1. Carga de datos**

````python
# Path of the file to read
flight_filepath = "./archivosDeDatos/flight_delays.csv"

# Read the file into a variable flight_data
flight_data = pd.read_csv(flight_filepath, index_col="Month")

````

### **04.3.3. Grafico de barras**

Los comandos para personalizar el texto (título y etiqueta del eje vertical) y el tamaño de la figura ya los conocemos del tutorial anterior. El código que crea el gráfico de barras es nuevo:

```python
# Gráfico de barras que muestra el retraso medio de llegada de los vuelos de Spirit Airlines por mes
sns.barplot(x=flight_data.index, y=flight_data[“NK”])
```

Consta de tres componentes principales:

- `sns.barplot`: indica al cuaderno que queremos crear un gráfico de barras.
  - Recuerda que *''sns''* hace referencia al paquete seaborn, y todos los comandos que utilices para crear gráficos en este curso comenzarán con este prefijo.
- `x=flight_data.index`: determina qué se utilizará en el eje horizontal. En este caso, hemos seleccionado la columna que indexa las filas (en este caso, la columna que contiene los meses).
- `y=flight_data[“NK”]`: establece la columna de los datos que se utilizará para determinar la altura de cada barra. En este caso, seleccionamos la columna *''NK'"*.

> **Nota importante**: Debes seleccionar la columna de indexación con flight_data.index, y no es posible utilizar *flight_data[“Month”]* (lo cual generará un error). Esto se debe a que, al cargar el conjunto de datos, se utilizó la columna *''Month''* para indexar las filas. Siempre debemos utilizar esta notación especial para seleccionar la columna de indexación.


````python
# Set the width and height of the figure
plt.figure(figsize=(10,6))

# Add title
plt.title("Average Arrival Delay for Spirit Airlines Flights, by Month")

# Bar chart showing average arrival delay for Spirit Airlines flights by month
sns.barplot(x=flight_data.index, y=flight_data['NK'])

# Add label for vertical axis
plt.ylabel("Arrival delay (in minutes)")
````

### **04.3.4. Mapas de calor**

El código correspondiente para crear el mapa de calor es el siguiente:

```python
# Mapa de calor que muestra el retraso medio de llegada de cada aerolínea por mes
sns.heatmap(data=flight_data, annot=True)
```
Este código tiene tres componentes principales:

- `sns.heatmap`: indica al cuaderno que queremos crear un mapa de calor.
- `data=flight_data`: indica al cuaderno que utilice todas las entradas de *flight_data* para crear el mapa de calor.
- `annot=True`: garantiza que los valores de cada celda aparezcan en el gráfico. (¡Si se omite, se eliminan los números de cada una de las celdas!)

¿Qué patrones puedes detectar en la tabla? Por ejemplo, si te fijas bien, los meses hacia finales de año (especialmente los meses 9 a 11) aparecen relativamente oscuros para todas las aerolíneas. ¡Esto sugiere que las aerolíneas son mejores (de media) a la hora de cumplir con los horarios durante estos meses!

````python
# Set the width and height of the figure
plt.figure(figsize=(14,7))

# Add title
plt.title("Average Arrival Delay for Each Airline, by Month")

# Heatmap showing average arrival delay for each airline by month
sns.heatmap(data=flight_data, annot=True)

# Add label for horizontal axis
plt.xlabel("Airline")
````

## **04.4. Graficos de dispersion**

### **04.4.1. Lectura de datos y primera visualizacoin de estos**

````python
# Path of the file to read
insurance_filepath = "./archivosDeDatos/insurance.csv"

# Read the file into a variable insurance_data
insurance_data = pd.read_csv(insurance_filepath)
````

````python
insurance_data.head()
````

### **04.4.2. Graficos de dispersion**

Para crear un gráfico de dispersión sencillo, utilizamos el comando `sns.scatterplot()` y especificamos los valores de:

- el eje horizontal x *(x=insurance_data[“bmi”])*, y
- el eje vertical y *(y=insurance_data[“charges”])*.


````python
sns.scatterplot(x=insurance_data['bmi'], y=insurance_data['charges'])
````

El gráfico de dispersión anterior sugiere que el índice de masa corporal (IMC o BMI en ingles) y las primas del seguro están correlacionados positivamente, es decir, que los clientes con un IMC más alto suelen pagar también más en concepto de primas de seguro. (Este patrón tiene sentido, ya que un IMC elevado suele estar asociado a un mayor riesgo de padecer enfermedades crónicas).

Para comprobar la solidez de esta relación, quizá te interese añadir una **línea que marca la regresion lineal**, es decir, la línea que mejor se ajusta a los datos. Para ello, cambiamos el comando por `sns.regplot()`.

> La regresión lineal es una técnica de análisis utilizada para predecir el valor de una variable dependiente basada en el valor de una o más variables independientes. La variable que desea predecir se denomina variable dependiente. La variable que está utilizando para predecir el valor de la otra variable se denomina variable independiente.

````python
sns.regplot(x=insurance_data['bmi'], y=insurance_data['charges'])
````

### **04.4.3. Graficos de dispersion con codigos de color**

Se introduce a *sns.scatterplot()* el valor de  `hue`el cual permite crear variables con color para su mas facil representacion.

````python
sns.scatterplot(x=insurance_data['bmi'], y=insurance_data['charges'], hue=insurance_data['smoker'])
````

Tambien se le puede asociar el valor de `hue` a *sns.lmplot()* y funciona exactamente igual que en el anterior ejemplo.

El comando `sns.lmplot()` siguiente funciona de forma ligeramente diferente a los comandos que se han visto hasta ahora:

- En lugar de establecer *x=insurance_data[“bmi”]* para seleccionar la columna *''bmi''* en insurance_data, establecemos *x="bmi"* para especificar únicamente el nombre de la columna.
- Del mismo modo, *y="charges"* y *hue="smoker"* también contienen los nombres de las columnas.
- Especificamos el conjunto de datos con *data=insurance_data*.

````python
sns.lmplot(x="bmi", y="charges", hue="smoker", data=insurance_data)
````

Tambien existe la siguiente representacion de los datos con la libreria `seaborn`.

````python
sns.swarmplot(x=insurance_data['smoker'],
              y=insurance_data['charges'])
````

## **04.5. Distribuciones**

### **04.5.1. Lectura incial de datos y prepacion del entorno**

````python


# Path of the file to read
iris_filepath = "./archivosDeDatos/iris.csv"

# Read the file into a variable iris_data
iris_data = pd.read_csv(iris_filepath, index_col="Id")

# Print the first 5 rows of the data
iris_data.head()

````

### **04.5.2. Histogramas**

````python
sns.histplot(iris_data['Petal Length (cm)'])
````

### **04.5.3. Graficos de densidad**

El siguiente tipo de gráfico es un gráfico de estimación de densidad por kernel (KDE-> kernel density estimate). Si no estás familiarizado con los gráficos KDE, puedes imaginártelos como un histograma suavizado.

Para crear un gráfico KDE, utilizamos el comando `sns.kdeplot()`. Al establecer shade=True, se colorea el área situada debajo de la curva (y data= permite seleccionar la columna que queremos representar).

````python
sns.kdeplot(data=iris_data['Petal Length (cm)'], shade=True)
````

### **04.5.4. Graficos KDE 2D**

A la hora de crear un gráfico KDE, no estamos limitados a una sola columna. Podemos crear un gráfico KDE bidimensional (2D) con el comando sns.jointplot.

En el gráfico que se muestra a continuación, la codificación por colores nos indica la probabilidad de que se den diferentes combinaciones de anchura de sépalo y longitud de pétalo, siendo las partes más oscuras de la figura las que tienen mayor probabilidad.

Ten en cuenta que, además del gráfico KDE en 2D situado en el centro,

- la curva situada en la parte superior de la figura es un gráfico KDE de los datos del eje x (en este caso, *iris_data['Petal Lenght (cm)'])*, y
- la curva situada a la derecha de la figura es un gráfico KDE de los datos del eje y (en este caso, *iris_data['Sepal Width (cm)']*).



````python
# 2D KDE plot
sns.jointplot(x=iris_data['Petal Length (cm)'], y=iris_data['Sepal Width (cm)'], kind="kde")
````

### **04.5.5. Graficos con codigos de color**

En la siguiente parte del tutorial, crearemos gráficos para comprender las diferencias entre las especies.

Podemos crear tres histogramas diferentes (uno para cada especie) de la longitud de los pétalos utilizando el comando `sns.histplot()` (como se ha visto anteriormente).

- `data=`: proporciona el nombre de la variable que hemos utilizado para leer los datos
- `x=`: establece el nombre de la columna con los datos que queremos representar gráficamente
- `hue=`: establece la columna que utilizaremos para dividir los datos en diferentes histogramas

````python
# Histograms for each species
sns.histplot(data=iris_data, x='Petal Length (cm)', hue='Species')

# Add title
plt.title("Histogram of Petal Lengths, by Species")
````

También podemos crear un gráfico KDE para cada especie utilizando `sns.kdeplot()` (como se ha hecho anteriormente). La funcionalidad de los parámetros ''data'', ''x'' ''hue'' es idéntica a la utilizada anteriormente con *sns.histplot*. Además, establecemos `shade=True` para colorear el área situada debajo de cada curva.

````python
# KDE plots for each species
sns.kdeplot(data=iris_data, x='Petal Length (cm)', hue='Species', shade=True)

# Add title
plt.title("Distribution of Petal Lengths, by Species")
````
